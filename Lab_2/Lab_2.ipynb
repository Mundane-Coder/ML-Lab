```python
import pandas as pd

dataset = pd.read_csv("enjoysport.csv")
print(dataset.head())
```

```python
ans = ['%', '%', '%', '%', '%', '%']

for row in dataset.values:
    if(row[len(row) - 1].upper() == "YES"):
        for i in range(len(row) - 1):
            if(ans[i] == '%'):
                ans[i] = row[i]
            elif ans[i] != row[i]:
                ans[i] = '?'
        
print(ans)
```
