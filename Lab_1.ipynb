```python
import numpy as np

class Adaline:

    def __init__(self, n_iterations, learning_rate, x, y):
        self.interations = n_iterations
        self.learning_rate = learning_rate
        X = np.array(x)
        y = np.array(y)



    def train (self):
        self.weights = np.zeros(1 + X.shape[1])
        self.weights[0] = 0
        
        for _ in range(self.interations):

            my_y = np.dot(X, self.weights[1:]) + self.weights[0]
            error = y - my_y

            self.weights[1:] += self.learning_rate * np.dot(X.T, error)
            self.weights[0] += self.learning_rate * np.sum(error)

    def predict(self, x):

        output = np.dot(x, self.weights[1:]) + self.weights[0]
        return output


X = np.array([[1, 2], [2, 3], [3, 4]])
Y = np.array([1, 1, 0])


ada = Adaline(10, 0.01, X, Y)
ada.train()

print("the output for [2. 3]", ada.predict([2, 3]))
            
```

```python
import numpy as np 

class Adaline:
    def __init__(self, learning_rate=0.01, n_iterations=100):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.bias = None
    
    def fit(self, X, y):
        # Initialize weights and bias randomly
        self.weights = np.zeros(1 + X.shape[1]) 
        self.bias = 0 
        
        for _ in range(self.n_iterations):
            net_input = np.dot(X, self.weights[1:]) + self.weights[0] 
            predicted = net_input
            errors = y - predicted
            
            # Update weights using the delta rule
            self.weights[1:] += self.learning_rate * np.dot(X.T, errors) 
            self.weights[0] += self.learning_rate * np.sum(errors)
    
    def predict(self, X):
        net_input = np.dot(X, self.weights[1:]) + self.weights[0]
        return net_input 
    
# Example usage
X = np.array([[1, 2], [2, 3], [3, 4]])
y = np.array([1, 1, 0])

adaline = Adaline(learning_rate=0.01, n_iterations=10)
adaline.fit(X, y)
print("Weights:", adaline.weights)
print("Prediction for [2, 3]:", adaline.predict(np.array([2, 3])))
```
