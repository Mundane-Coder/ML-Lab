```python
import csv
import pandas as pd

# [Color: #FF8C00] Define the raw data array
csv_data = """text,label
This is the first document about technology.,technology
Another document discussing AI and machine learning.,technology
A sports article about football.,sports
A news report on the latest cricket match.,sports
Document related to politics and elections.,politics
A piece about the current economic situation.,economics
The new smartphone features a high-resolution camera.,technology
Quantum computing could revolutionize cryptography.,technology
The tennis championship final was a thrilling match.,sports
Olympic athletes train for years to win a medal.,sports
The parliament passed a new law regarding taxation.,politics
The prime minister addressed the nation on television.,politics
Inflation rates have hit a record high this quarter.,economics
The stock market crashed after the unexpected news.,economics
Software engineers are developing a new operating system.,technology
The local basketball team won their regional tournament.,sports
Citizens are preparing to cast their votes in the upcoming election.,politics
Global trade policies are shifting due to tariffs.,economics
Artificial intelligence is generating new forms of art.,technology
The marathon runner broke the world record today.,sports
The senate debated the healthcare bill for hours.,politics
Interest rates were lowered by the central bank.,economics
Virtual reality headsets are becoming more affordable.,technology
The soccer World Cup draws billions of viewers.,sports
Diplomatic relations between the two countries improved.,politics
Small businesses are struggling with the new economic reforms.,economics
Cloud computing architecture relies on distributed servers.,technology"""

with open('example_csv.csv', 'w') as file:
          file.write(csv_data)

values = pd.read_csv('example_csv.csv')
print(values)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report
from sklearn.feature_extraction.text import TfidfVectorizer

texts = values['text']
labels = values['label']

tfidf = TfidfVectorizer()
X = tfidf.fit_transform(texts)

x_train, x_test, y_train, y_test = train_test_split(X, labels, test_size = 0.2, random_state = 42)

classifier = MultinomialNB()

classifier.fit(x_train, y_train)

y_pred = classifier.predict(x_test)

accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average = 'weighted')
recall = recall_score(y_test, y_pred, average = 'weighted')
report = classification_report(y_test, y_pred)

print(accuracy, precision, recall, report)
```

```python
new_texts = ["This is a new document about technology.", "Another document about sports."]

new_data = tfidf.transform(new_texts)
y_pred = classifier.predict(new_data)
print(y_pred)
```
