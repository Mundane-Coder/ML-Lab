```python
import pandas as pd
import numpy as np


data_file = pd.read_csv("enjoysport.csv")
print(data_file.head())
```

```python
rows = []
target = []

for row in data_file.values:
    rows.append(row[:-1])
    target.append(row[-1])

g = []
s = rows[0].copy()

for x in range(len(s)):
    temp = []
    for y in range(len(s)):
        temp.append('?')
    g.append(temp)


for i, row in enumerate(rows):
    # FIX 1: Look at the current target label, not the last one
    if target[i].upper() == "YES":
        for idx in range(len(s)):
            if s[idx] != row[idx]:
                s[idx] = '?'
        # (Remember: A full algorithm also eliminates inconsistent G hypotheses here)
                
    else:
        for idx in range(len(s)):
            # FIX 2 & 3: We compare the specific attribute index in G
            # and use standard Python list indexing syntax.
            if s[idx] != row[idx]:
                g[idx][idx] = s[idx]
            else:
                g[idx][idx] = '?'

g = np.array(g)
s = np.array(s)

print("Final G:", g)
print("Final S:", s)
        
```
