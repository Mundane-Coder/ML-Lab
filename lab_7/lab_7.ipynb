```python
import csv
import pandas as pd
import numpy as np

csv_data = ''' 

age,sex,cp,trestbps,chol,fbs,restecg,thalach,exang,oldpeak,slope,ca,thal,heartdisease
63,1,1,145,233,1,2,150,0,2.3,3,0,6,0
67,1,4,160,286,0,2,108,1,1.5,2,3,3,2
67,1,4,120,229,0,2,129,1,2.6,2,2,7,1
37,1,3,130,250,0,0,187,0,3.5,3,0,3,0
41,0,2,130,204,0,2,172,0,1.4,1,0,3,0
62,0,4,140,268,0,2,160,0,3.6,3,2,3,3
60,1,4,130,206,0,2,132,1,2.4,2,2,?,4

'''


with open('data_csv.csv', 'w') as file:
    file.write(csv_data)

heartDisease = pd.read_csv('data_csv.csv')
print(heartDisease)
heartDisease.replace('?', np.nan)
```

```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.inference import VariableElimination


model = DiscreteBayesianNetwork([
    ('age', 'trestbps'),
    ('age', 'fbs'),
    ('sex', 'trestbps'),
    ('exang', 'trestbps'),
    ('trestbps', 'heartdisease'),
    ('fbs', 'heartdisease'),
    ('heartdisease', 'thalach'),
    ('heartdisease', 'chol')
])


#model.fit(heartDisease, estimator = DiscreteMLE)
model.fit(heartDisease)
heartDisease_infer = VariableElimination(model)

q1 = heartDisease_infer.query(variables = ['heartdisease'], evidence = {'age' : 37})
print(q1)
```
